Podemos observar que el porcentaje de datos nulos en cada columna son menores al 4%, por ende voy a proceder a llenar estos espacios con la mediana de el resto, usare la mediana porque no se desplaza por datos atipicos como la media.

```python
# =========================================================
# 1. ELIMINAR FILAS SIN VARIABLE OBJETIVO
# =========================================================

df_completos = df_completos.dropna(subset=["temp_max_manana"]).copy()


# =========================================================
# 2. CORREGIR Y APROVECHAR LA FECHA
# =========================================================

df_completos["fecha"] = pd.to_datetime(
    df_completos["fecha"],
    errors="coerce"
)

# Eliminar únicamente filas cuya fecha no se pueda interpretar
df_completos = df_completos.dropna(subset=["fecha"])


# =========================================================
# 3. RECONSTRUIR VARIABLES DE FECHA
# =========================================================

# Año correcto
df_completos["anio"] = df_completos["fecha"].dt.year

# Día correcto del año
df_completos["dia_del_anio"] = df_completos["fecha"].dt.dayofyear

# Mes correcto
df_completos["mes"] = (
    df_completos["fecha"]
    .dt.month_name()
    .str.lower()
)

# Estación correcta según el mes
estaciones = {
    12: "invierno",
    1: "invierno",
    2: "invierno",

    3: "primavera",
    4: "primavera",
    5: "primavera",

    6: "verano",
    7: "verano",
    8: "verano",

    9: "otono",
    10: "otono",
    11: "otono"
}

df_completos["estacion_anio"] = (
    df_completos["fecha"].dt.month.map(estaciones)
)


# =========================================================
# 4. CORREGIR HUMEDAD
# =========================================================

columnas_humedad = [
    "humedad_media",
    "humedad_min",
    "humedad_max"
]

for columna in columnas_humedad:

    # Algunos valores están entre 0 y 1 en vez de 0 y 100
    mascara = (
        (df_completos[columna] >= 0) &
        (df_completos[columna] <= 1)
    )

    df_completos.loc[mascara, columna] = (
        df_completos.loc[mascara, columna] * 100
    )

    # Humedades imposibles se convierten en nulos
    df_completos.loc[
        (df_completos[columna] < 0) |
        (df_completos[columna] > 100),
        columna
    ] = float("nan")


# =========================================================
# 5. CORREGIR PRESIÓN
# =========================================================

columnas_presion = [
    "presion_media",
    "presion_min",
    "presion_max"
]

for columna in columnas_presion:

    df_completos.loc[
        (df_completos[columna] < 900) |
        (df_completos[columna] > 1100),
        columna
    ] = float("nan")


# =========================================================
# 6. CORREGIR VARIABLES QUE NO PUEDEN SER NEGATIVAS
# =========================================================

columnas_no_negativas = [
    "presion_desv",
    "humedad_desv",
    "viento_media",
    "viento_min",
    "viento_max",
    "viento_desv",
    "rafaga_media",
    "rafaga_min",
    "rafaga_max",
    "rafaga_desv"
]

for columna in columnas_no_negativas:

    df_completos.loc[
        df_completos[columna] < 0,
        columna
    ] = float("nan")


# =========================================================
# 7. CORREGIR DIRECCIÓN DEL VIENTO
# =========================================================

df_completos.loc[
    (df_completos["direccion_viento"] < 0) |
    (df_completos["direccion_viento"] >= 360),
    "direccion_viento"
] = float("nan")


# =========================================================
# 8. RECONSTRUIR SECTOR DEL VIENTO
# =========================================================

def obtener_sector(angulo):

    if pd.isna(angulo):
        return None

    angulo = angulo % 360

    sectores = [
        "N",
        "NE",
        "E",
        "SE",
        "S",
        "SO",
        "O",
        "NO"
    ]

    indice = int(((angulo + 22.5) // 45) % 8)

    return sectores[indice]


df_completos["sector_viento"] = (
    df_completos["direccion_viento"].apply(obtener_sector)
)

# =========================================================
# 9. NORMALIZAR OTRAS COLUMNAS DE TEXTO
# =========================================================

columnas_texto = df_completos.select_dtypes(
    include=["object", "string"]
).columns

for columna in columnas_texto:

    # Convertir a object para evitar pd.NA
    df_completos[columna] = df_completos[columna].astype("object")

    # Reemplazar valores faltantes por NaN estándar
    df_completos[columna] = df_completos[columna].where(
        df_completos[columna].notna(),
        float("nan")
    )

    # Limpiar espacios y pasar a minúsculas
    df_completos[columna] = (
        df_completos[columna]
        .str.strip()
        .str.lower()
    )


# =========================================================
# 10. ELIMINAR FECHA
# =========================================================

# Ya utilizamos la fecha para obtener mes, año y día del año.
# No la utilizaremos directamente en LinearRegression.

df_completos = df_completos.drop(
    columns=["fecha"],
    errors="ignore"
)


# =========================================================
# 11. VERIFICACIÓN
# =========================================================

print("Dimensiones finales:")
print(df_completos.shape)

print("\nValores nulos restantes:")
print(df_completos.isnull().sum())

print("\nTotal de valores nulos:")
print(df_completos.isnull().sum().sum())
```

Modelo:

```python
X = df_completos.drop(columns=["temp_max_manana"])
y = df_completos["temp_max_manana"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

```python
X_train = X_train.copy()
X_test = X_test.copy()

columnas_atipicos = X_train.select_dtypes(include="number").columns.tolist()

columnas_excluir = [
    "anio",
    "dia_del_anio",
    "direccion_viento"
]

columnas_atipicos = [
    columna for columna in columnas_atipicos
    if columna not in columnas_excluir
]

limites_atipicos = {}

for columna in columnas_atipicos:

    Q1 = X_train[columna].quantile(0.25)
    Q3 = X_train[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 3 * IQR
    limite_superior = Q3 + 3 * IQR

    limites_atipicos[columna] = (
        limite_inferior,
        limite_superior
    )

    mascara_train = (
        (X_train[columna] < limite_inferior) |
        (X_train[columna] > limite_superior)
    )

    cantidad = mascara_train.sum()

    print(
        columna,
        "- Atípicos:",
        cantidad
    )

    X_train.loc[
        mascara_train,
        columna
    ] = float("nan")

    mascara_test = (
        (X_test[columna] < limite_inferior) |
        (X_test[columna] > limite_superior)
    )

    X_test.loc[
        mascara_test,
        columna
    ] = float("nan")
```

```python
columnas_numericas = X_train.select_dtypes(include="number").columns
columnas_categoricas = X_train.select_dtypes(exclude="number").columns

print("Numéricas:")
print(list(columnas_numericas))

print("\nCategóricas:")
print(list(columnas_categoricas))
```

```python
pipeline_numerico = Pipeline(
    steps=[
        ("imputacion", SimpleImputer(strategy="median"))
    ]
)

pipeline_categorico = Pipeline(
    steps=[
        ("imputacion", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocesador = ColumnTransformer(
    transformers=[
        ("numericas", pipeline_numerico, columnas_numericas),
        ("categoricas", pipeline_categorico, columnas_categoricas)
    ]
)
```

```python
modelo_1 = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador),
        ("regresion", LinearRegression())
    ]
)
```

```python
modelo_1.fit(X_train, y_train)
```

```python
y_pred = modelo_1.predict(X_test)
```

```python
rmse = mean_squared_error(y_test, y_pred) ** 0.5
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("RMSE:", rmse)
print("MAE:", mae)
print("R²:", r2)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Temperatura real")
plt.ylabel("Temperatura predicha")
plt.title("Valores reales vs predicciones")

plt.show()
```

```python
residuos = y_test - y_pred

plt.figure(figsize=(8, 6))

plt.scatter(y_pred, residuos, alpha=0.5)

plt.axhline(y=0)

plt.xlabel("Valores predichos")
plt.ylabel("Residuos")
plt.title("Residuos del modelo")

plt.show()
```
